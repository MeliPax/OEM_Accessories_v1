```python
# split year_range (e.g. "2019-2021") into min_year and max_year columns
def split_year_range(df):
    df = df.copy()
    years = df['year_range'].str.split('-', n=1, expand=True)

    df['min_year'] = pd.to_numeric(years[0].str.strip(), errors='coerce').astype('Int64')
    # if there is no "-", max_year is the same as min_year
    df['max_year'] = pd.to_numeric(years[1].str.strip(), errors='coerce').astype('Int64').fillna(df['min_year'])
    return df

df_essential = split_year_range(df_essential)
```

```python
# business rules for the data quality check (source column names are mapped to the renamed ones via column_mapping)
data_rules = {
    'greater_than':         [('MSRP', 'cost')],                # left value must be > right value
    'greater_or_equal':     [('max_year', 'min_year')],        # left value must be >= right value
    'not_null':             ['stockCode', 'model', 'productDescription', 'MSRP', 'cost', 'laborTime'],
    'numeric_non_zero':     ['msrp', 'cost', 'min_year', 'max_year'],
    'numeric_non_negative': ['labor_time'],                    # zero or more
}

def data_quality_check(df, data_rules, column_mapping=column_mapping):
    rename = {src: new for src, (new, _) in column_mapping.items()}
    col = lambda c: rename.get(c, c)

    results = []      # one row per rule check
    failed_rows = {}  # rule description -> rows that broke it

    def record(rule, columns, mask=None, note=''):
        n_failed = int(mask.sum()) if mask is not None else None
        status = 'MISSING COLUMN' if mask is None else ('PASS' if n_failed == 0 else 'FAIL')
        results.append({'rule': rule, 'columns': columns, 'status': status, 'failed_rows': n_failed, 'note': note})
        if mask is not None and n_failed:
            failed_rows[f"{rule}: {columns}"] = df[mask]

    def missing(*cols):
        return [c for c in cols if c not in df.columns]

    for rule in ('greater_than', 'greater_or_equal'):
        for left, right in data_rules.get(rule, []):
            l, r = col(left), col(right)
            label = f"{l} {'>' if rule == 'greater_than' else '>='} {r}"
            if missing(l, r):
                record(rule, label, note=f"not in df: {missing(l, r)}")
                continue
            ok = df[l] > df[r] if rule == 'greater_than' else df[l] >= df[r]
            record(rule, label, ~ok.fillna(False).astype(bool))

    for c in data_rules.get('not_null', []):
        if missing(col(c)):
            record('not_null', col(c), note=f"'{c}' not in df")
            continue
        record('not_null', col(c), df[col(c)].isna())

    # numeric rules: a value fails if it is not a number, or if it breaks the condition
    numeric_fails = {
        'numeric_non_zero':     lambda v: v == 0,
        'numeric_non_negative': lambda v: v < 0,
    }
    for rule, fails in numeric_fails.items():
        for c in data_rules.get(rule, []):
            if missing(col(c)):
                record(rule, col(c), note=f"'{c}' not in df")
                continue
            values = pd.to_numeric(df[col(c)], errors='coerce')
            record(rule, col(c), (values.isna() | fails(values)).astype(bool))

    report = pd.DataFrame(results)
    broken = report['status'].ne('PASS').sum()
    print(f"Data quality check: {len(report)} checks, {broken} with issues")
    return report, failed_rows
```

```python
# dict: model_name -> df with one record per applicable year (min_year to max_year, inclusive)

output_columns = ['part_number', 'model_name', 'product_description', 'comments',
                  'msrp', 'cost', 'labor_time', 'min_year', 'max_year', 'applicable_year']

def build_model_year_dict(df):
    df = df.copy()
    # list of every year in the range, then one row per year
    df['applicable_year'] = [list(range(int(lo), int(hi) + 1)) for lo, hi in zip(df['min_year'], df['max_year'])]
    df = df.explode('applicable_year')
    df['applicable_year'] = df['applicable_year'].astype(int)

    return {
        model: group[output_columns].sort_values(['part_number', 'applicable_year']).reset_index(drop=True)
        for model, group in df.groupby('model_name')
    }
```

```python
# show the records whose Description matches the given value (case-insensitive, ignores leading/trailing spaces)
def show_records_by_description(df, description, column='Description', partial=False, exclude=None,
                                years=None, year_column='ModelYear'):
    values = df[column].astype(str).str.strip().str.lower()
    target = description.strip().lower()

    # partial=True matches descriptions that contain the value instead of an exact match
    mask = values.str.contains(target, regex=False) if partial else values.eq(target)

    # exclude: list of sub-strings, any record whose value contains one of them is dropped
    for sub in exclude or []:
        mask &= ~values.str.contains(sub.strip().lower(), regex=False)

    # years: a single year or a list/range of years, only records in those years are kept
    if years is not None:
        years = [years] if isinstance(years, (int, str)) else list(years)
        years = [int(y) for y in years]
        mask &= pd.to_numeric(df[year_column], errors='coerce').isin(years)

    matches = df[mask].sort_values(year_column) if year_column in df.columns else df[mask]

    excluded_note = f", excluding {exclude}" if exclude else ""
    years_note = f", years {years}" if years is not None else ""
    print(f"{len(matches)} record(s) with {column} {'containing' if partial else '='} '{description}'{excluded_note}{years_note}")
    return matches
```

```python
import re

# alternate spellings in the VW models data -> the name used in the accessories data
VW_DESCRIPTION_ALIASES = {
    "Atlas Sport Cross": "Atlas Cross Sport",   # 2020-2023 Atlas Cross Sport records
}

# replace each alias phrase with its standard name (case-insensitive, whole words only)
def apply_aliases(value, aliases):
    if not isinstance(value, str) or not aliases:
        return value
    for alias, standard in aliases.items():
        value = re.sub(rf"\b{re.escape(alias)}\b", standard, value, flags=re.IGNORECASE)
    return value
```

```python
# clean a text column into a new column (<column>_clean by default):
# replace alias spellings (optional), trim spaces on both ends, collapse inner spaces, and drop repeated words
# e.g. "Jetta GLI GLI" -> "Jetta GLI", "ID. Buzz ID. Buzz 1st Edition" -> "ID. Buzz 1st Edition"
# aliases: optional dict of alternate spelling -> standard name, e.g. VW_DESCRIPTION_ALIASES

def clean_text_column(df, column, new_column=None, aliases=None):
    df = df.copy()
    new_column = new_column or f"{column}_clean"

    def remove_repeated_words(value):
        if not isinstance(value, str):
            return value
        seen = set()
        words = []
        for word in value.strip().split():
            # compare case-insensitively, keep the first occurrence as written
            if word.lower() not in seen:
                seen.add(word.lower())
                words.append(word)
        return ' '.join(words)

    # aliases first, so any repeated words they create are removed too
    df[new_column] = df[column].map(lambda v: remove_repeated_words(apply_aliases(v, aliases)))

    changed = (df[column].astype(str) != df[new_column].astype(str)).sum()
    print(f"Cleaned {column} -> {new_column}: {changed} of {len(df)} values changed")
    return df
```

```python
# check how many unique model names in the accessories data are a total (exact) match
# to a description in the VW models data (case-insensitive, ignores leading/trailing spaces)
def match_model_names(accessories_df, models_df, model_column='model_name', description_column='Description_clean'):
    normalize = lambda s: s.astype(str).str.strip().str.lower()

    # number of VW model records for each normalized description
    description_counts = normalize(models_df[description_column]).value_counts()

    results = pd.DataFrame({model_column: sorted(accessories_df[model_column].dropna().unique())})
    results['vw_model_records'] = normalize(results[model_column]).map(description_counts).fillna(0).astype(int)
    results['matched'] = results['vw_model_records'] > 0
    results = results.sort_values(['matched', model_column], ascending=[False, True]).reset_index(drop=True)

    total = len(results)
    matched = int(results['matched'].sum())
    print(f"Unique {model_column} values: {total}")
    print(f"Total match in {description_column}: {matched} ({matched / total:.0%})")
    print(f"No match: {total - matched}")
    display(results)
    return results

```

```python
# for each unique model name in the accessories data, list the description variations in the VW models data
# that contain it (partial, case-insensitive), one variation per row under its model name
# exclude_longer_names=True drops descriptions that belong to a longer model name in the same list
#   e.g. "Golf" skips "Golf R ..." and "e-Golf ...", "Atlas" skips "Atlas Cross Sport ..."
# exclude: optional dict of extra sub-strings to drop per model, e.g. {"Golf": ["GTI"]}
# output columns, prefixed with the dataset they come from:
#   accessory_model      model name from the accessories data (accessories_df)
#   vw_variations_count  number of distinct variations found in the VW models data (vw_models_df)
#   vw_variation         description variation from the VW models data
#   vw_model_numbers     distinct model numbers of the VW models data rows with that variation
#   vw_model_records     number of VW models data rows with that variation
#   vw_model_years       model years of those VW models data rows
def find_model_variations(accessories_df, models_df, model_column='model_name', description_column='Description_clean',
                          year_column='ModelYear', model_number_column='ModelNumber', exclude_longer_names=True, exclude=None):
    model_names = sorted(accessories_df[model_column].dropna().astype(str).str.strip().unique())
    descriptions = models_df[description_column].astype(str).str.strip()
    lowered = descriptions.str.lower()
    exclude = exclude or {}

    rows = []
    for name in model_names:
        mask = lowered.str.contains(name.lower(), regex=False)

        skip = list(exclude.get(name, []))
        if exclude_longer_names:
            skip += [other for other in model_names if other != name and name.lower() in other.lower()]
        for sub in skip:
            mask &= ~lowered.str.contains(sub.strip().lower(), regex=False)

        matches = models_df[mask].assign(_variation=descriptions[mask])
        if matches.empty:
            rows.append({'accessory_model': name, 'vw_variations_count': 0, 'vw_variation': '(no match)',
                         'vw_model_numbers': '', 'vw_model_records': 0, 'vw_model_years': ''})
            continue

        variations_count = matches['_variation'].nunique()
        for variation, group in matches.groupby('_variation'):
            years = sorted(group[year_column].dropna().astype(int).unique())
            model_numbers = sorted(group[model_number_column].dropna().astype(str).str.strip().unique())
            rows.append({
                'accessory_model': name,
                'vw_variations_count': variations_count,
                'vw_variation': variation,
                'vw_model_numbers': ', '.join(model_numbers),
                'vw_model_records': len(group),
                'vw_model_years': ', '.join(str(y) for y in years),
            })

    results = pd.DataFrame(rows)

    per_model = results.groupby('accessory_model')['vw_variations_count'].first()
    print(f"Unique {model_column} values: {len(per_model)}")
    print(f"With at least one variation: {(per_model > 0).sum()}, without: {(per_model == 0).sum()}")
    print(f"Total variations: {per_model.sum()}")

    # show the model name and count once per model, with its variations listed vertically beneath it
    view = results.copy()
    first_row = ~view['accessory_model'].duplicated()
    view['vw_variations_count'] = view['vw_variations_count'].astype(str).where(first_row, '')
    view['accessory_model'] = view['accessory_model'].where(first_row, '')
    view.index = [''] * len(view)
    with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
        display(view)
    return results
```

```python
# one row per model: the unique keywords found across all of its VW variations, and how many there are
# takes the output of find_model_variations()
# words are split on spaces, wrapping punctuation is stripped (e.g. "*Ltd Avail*" -> "Ltd", "Avail"),
# duplicates are dropped case-insensitively, and the model name's own words are left out
# output columns:
#   accessory_model        model name from the accessories data
#   vw_variations_count    number of distinct variations found in the VW models data
#   vw_keywords_count      number of unique keywords across those variations
#   vw_variation_keywords  the unique keywords, sorted alphabetically
def extract_model_keywords(model_variations):
    rows = []
    for name, group in model_variations.groupby('accessory_model', sort=True):
        skip_words = {w.lower() for w in name.split()}
        keywords = {}
        for variation in group['vw_variation'].unique():
            if variation == '(no match)':
                continue
            for word in variation.split():
                word = word.strip('*(),')
                if word and word.lower() not in skip_words:
                    keywords.setdefault(word.lower(), word)

        sorted_keywords = sorted(keywords.values(), key=str.lower)
        rows.append({
            'accessory_model': name,
            'vw_variations_count': group['vw_variations_count'].iloc[0],
            'vw_keywords_count': len(sorted_keywords),
            'vw_variation_keywords': ', '.join(sorted_keywords),
        })

    model_keywords = pd.DataFrame(rows)
    print(f"Models: {len(model_keywords)}, unique keywords per model: {model_keywords['vw_keywords_count'].min()}-{model_keywords['vw_keywords_count'].max()}")
    with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
        display(model_keywords)
    return model_keywords
```
