```python

mazda_to_standard_columns_names = {
    'ModelYear': 'model_year',
    'CarLineCode': 'model',
    'TrimLevel': 'model_number',
    'PartNumber': 'part_number',
    'AccessoryName': 'part_description_en',
    'AccessoryNameFR': 'part_description_fr',
    'RestrictionEn': 'comments_en',
    'RestrictionFr': 'comments_fr',
    'MSRP': 'MSRP',
    'StandardInstallTime': 'install_hours'
}
```

```python
# change the column names to match the standard format
mazda_df.rename(columns=mazda_to_standard_columns_names, inplace=True)
```

```python
# What's the character lenght of the part number column? Is it consistent across all the rows? This is important to know because we will need to separate the part number and the package number, and we need to know how many characters to separate. We will assume that the last 4 characters of the part number column are the package number, and the rest are the part number. If this is not the case, we will need to find another way to separate them.
mazda_df['model_number_length'] = mazda_df['model_number'].str.len()
```

```python
mazda_df["model_number_length"].value_counts()
```

```python
# create two new columns, one named package and another one named long_model_number. Package columns will store the last 4 characters of the model_number, the long model_number will be a copy of the model_number column, and the model_number column will be truncated to the first characters minus the four last characters. This is because the model_number column in the mazda feed contains both the model number and the package number, and we want to separate them into two columns for better analysis. The long_model_number column will be useful for reference and for any future needs that may arise.

mazda_df['package'] = mazda_df['model_number'].str[-4:]
mazda_df['long_model_number'] = mazda_df['model_number']
mazda_df['short_model_number'] = mazda_df['model_number'].str[:-4]
```

```python
# Check if each value in package contains two letters and two numbers, which is the expected format for the package number. This function will be useful to identify any anomalies in the data and to ensure that the package number is correctly separated from the model number.

def check_package_format(package):
    if len(package) != 4:
        return False
    if not package[:2].isalpha():
        return False
    if not package[2:].isdigit():
        return False
    return True

# Then assign the result of this function to a new column named package_format_check, which will contain True for rows where the package number is in the correct format and False for rows where it is not. This will allow us to easily filter and analyze any rows that do not meet the expected format for the package number.
mazda_df['package_format_check'] = mazda_df['package'].apply(check_package_format)
```

```python
columns_mapper_for_rateImporter = {
    "model_year": "Year",
    "short_model_number": "Model",
    "package": "Package",
    "part_number": "Part",
    "part_description_en": "Description",
    "comments_en": "Comments",
    'part_description_fr': 'Description',
    'comments_fr': 'Comments',
    'MSRP': 'Price',
    'install_hours': 'Hours'
}

def standardize_column_names(df, column_mapping):
    """
    Standardize column names in df using a mapping dict {current_name: new_name}.
    Prints the changes made and returns the updated DataFrame.
    Handles missing columns by warning and skipping.
    """

    df = df.copy()
    original_columns = df.columns.tolist()
    renamed_columns = []
    missing_columns = []

    for current_name, new_name in column_mapping.items():
        if current_name in df.columns:
            df.rename(columns={current_name: new_name}, inplace=True)
            renamed_columns.append(f"'{current_name}' -> '{new_name}'")
        else:
            missing_columns.append(current_name)

    return df



def filter_for_rate_importer_columns_only(df):
    expected_columns = list(set(columns_mapper_for_rateImporter.values()))
    return df[expected_columns]

```

```python
# now to create the feed for each model, we will need to separate the data by model and then create a feed for each model. We can do this by using the groupby function in pandas to group the data by the model column, and then we can iterate through each group to create a feed for each model. This will allow us to have a separate feed for each model, which will be easier to manage and analyze. now, only add model numbers that only pass the package format check, because we want to make sure that the data is clean and accurate before we create the feeds. We can filter the data to include only rows where the package_format_check column is True, which will ensure that we are only including rows with valid package numbers in our feeds.

mazda_df_final = mazda_df[mazda_df['package_format_check'] == True]


```

```python

# Creating a model and trim specific feed for Mazda, which will be used for the integration with the dealer portal. This will create a dict that will contains {year: model: trim: dataframe} structure, which will be used to create the feed for each model and trim.

mazda_model_trim_feed = {} 

for model in mazda_df_final['model'].unique():
    model_df = mazda_df_final[mazda_df_final['model'] == model]
    mazda_model_trim_feed[model] = {}
    for year in model_df['model_year'].unique():
        year_df = model_df[model_df['model_year'] == year]
        mazda_model_trim_feed[model][year] = {}
        model_combined_df_EN = pd.DataFrame()
        model_combined_df_FR = pd.DataFrame()

        for trim in model_df['long_model_number'].unique():
            trim_df = model_df[model_df['long_model_number'] == trim]
            # display(trim_df.head())
            english_df = trim_df[["model_year", "model", "short_model_number", "package", 'part_number', 'part_description_en', 'comments_en', 'MSRP', 'install_hours']].reset_index(drop=True) 
            french_df = trim_df[["model_year", "model", "short_model_number", "package", 'part_number', 'part_description_fr', 'comments_fr', 'MSRP', 'install_hours']].reset_index(drop=True)
            english_df.index = english_df.index + 1
            french_df.index = french_df.index + 1

            # Change columns to match rateImporter
            english_df_columns_to_RateImporter = standardize_column_names(english_df, columns_mapper_for_rateImporter)
            french_df_columns_to_RateImporter = standardize_column_names(french_df, columns_mapper_for_rateImporter)
            
            model_combined_df_EN = pd.concat([model_combined_df_EN, english_df_columns_to_RateImporter], ignore_index=True)
            model_combined_df_FR = pd.concat([model_combined_df_FR, french_df_columns_to_RateImporter], ignore_index=True)

            ready_to_upload_df_EN = model_combined_df_EN[['Year', 'Model', 'Package', 'Part', 'Description', 'Price',  'Hours', 'Comments']]
            ready_to_upload_df_FR = model_combined_df_FR[['Year', 'Model', 'Package', 'Part', 'Description', 'Price',  'Hours', 'Comments']]
        mazda_model_trim_feed[model] = {
            "en": ready_to_upload_df_EN,
            "fr": ready_to_upload_df_FR
        }
```

```python

def save_to_excel_v1(data_dict, file_path):
    # with pd.ExcelWriter(file_path) as writer:
    for year, models in data_dict.items():
        for model, trims in models.items():
            english_df = pd.DataFrame()
            french_df = pd.DataFrame()
            for trim, languages in trims.items():
                english_df = pd.concat([english_df, languages["en"]], ignore_index=True)
                french_df = pd.concat([french_df, languages["fr"]], ignore_index=True)
            english_sheet_name = f"{year}_{model}_en"
            french_sheet_name = f"{year}_{model}_fr"

            return english_df, french_df
            # english_df.to_excel(writer, sheet_name=english_sheet_name, index=False)
            # french_df.to_excel(writer, sheet_name=french_sheet_name, index=False)


def save_to_excel_v2(data_dict, file_path):
    with pd.ExcelWriter(file_path) as writer:
        for model in data_dict.keys():
            english_df = data_dict[model]["en"]
            french_df = data_dict[model]["fr"]
            english_sheet_name = f"{model}_en"
            french_sheet_name = f"{model}_fr"
            english_df.to_excel(writer, sheet_name=english_sheet_name, index=False)
            french_df.to_excel(writer, sheet_name=french_sheet_name, index=False)


```

```python

# the name of the file will need to contain the time it was created, to make it easier to manage the files and to keep track of the different versions of the feed. We can use the datetime module in Python to get the current date and time, and then we can format it to include in the file name. This will allow us to have a clear and organized way to manage the different versions of the feed, and it will also make it easier to identify the most recent version of the feed when we need to use it for the integration with the dealer portal.

time = pd.Timestamp.now().strftime("%Y%m%d-%H%M%S")

# first check if the directory exists, if not create it
output_directory = pathlib.Path("../ready_to_upload/2026/Mazda")
output_directory.mkdir(parents=True, exist_ok=True)

mazda_ready_to_load_excel_file = output_directory / f"mazda_accy_ready_to_upload_{time}.xlsx"

save_to_excel_v2(mazda_model_trim_feed, mazda_ready_to_load_excel_file)
```
