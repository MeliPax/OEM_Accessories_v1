```python
import pandas as pd
from pathlib import Path

print("✓ Libraries loaded")
```

## STEP 1: Select ONE File to Explore

```python
# Find all Honda files
landing_zone = Path("../../../landing_zone/2026/Honda")
excel_files = sorted(list(landing_zone.rglob("*.xlsx")))

print(f"All Honda files:\n")
for i, file in enumerate(excel_files, 1):
    size_mb = file.stat().st_size / (1024**2)
    print(f"{i:2d}. {file.name:<50} ({size_mb:6.2f} MB)")

print(f"\n" + "="*80)
print("Which file should we explore? (Pick one by number or use the one below)")
print("="*80)
```

## STEP 2: Show ALL Sheets in File

```python
# Load Excel file
xl = pd.ExcelFile(test_file)
all_sheets = xl.sheet_names

print(f"File: {test_file.name}")
print(f"Total sheets: {len(all_sheets)}\n")

print("ALL SHEETS:")
print("="*80)
for i, sheet_name in enumerate(all_sheets, 1):
    print(f"{i:2d}. {sheet_name}")
```

## STEP 3: Look for EN and FR Sheets

Which sheets have EN (English) and which have FR (French) data?

```python
print("SEARCHING FOR EN/FR PATTERNS:\n")

# Look for different patterns
patterns = {
    "Contains '_EN'": [s for s in all_sheets if "_EN" in s],
    "Contains '_FR'": [s for s in all_sheets if "_FR" in s],
    "Contains '_APP_EN'": [s for s in all_sheets if "_APP_EN" in s],
    "Contains '_APP_FR'": [s for s in all_sheets if "_APP_FR" in s],
    "Contains 'EN'": [s for s in all_sheets if "EN" in s and "_" not in s],
    "Contains 'FR'": [s for s in all_sheets if "FR" in s and "_" not in s],
}

for pattern_name, sheets in patterns.items():
    if sheets:
        print(f"✓ {pattern_name}: {len(sheets)} sheets")
        for s in sheets:
            print(f"    • {s}")
    else:
        print(f"✗ {pattern_name}: None found")
    print()
```

## STEP 4: Show Data from Each Sheet (First Few Rows)

Let's look at what's inside each sheet

```python
# Show first 3 sheets in detail
sheets_to_inspect = all_sheets[:3]

for sheet_name in sheets_to_inspect:
    print(f"\n{'='*80}")
    print(f"Sheet: {sheet_name}")
    print(f"{'='*80}\n")
    
    try:
        df = pd.read_excel(test_file, sheet_name=sheet_name)
        print(f"Size: {len(df)} rows × {len(df.columns)} columns")
        print(f"\nColumns: {list(df.columns)}\n")
        print("First 5 rows:\n")
        display(df.head(5))
    except Exception as e:
        print(f"Error reading sheet: {str(e)}")
```

## STEP 6: Load the Sheets You Selected

```python
if not parts_sheet_en or not parts_sheet_fr:
    print("❌ Please edit Step 5 and specify the sheet names first")
else:
    print(f"Loading sheets...\n")
    
    try:
        df_en = pd.read_excel(test_file, sheet_name=parts_sheet_en)
        df_fr = pd.read_excel(test_file, sheet_name=parts_sheet_fr)
        
        print(f"✓ EN Sheet loaded: {len(df_en)} rows × {len(df_en.columns)} columns")
        print(f"✓ FR Sheet loaded: {len(df_fr)} rows × {len(df_fr.columns)} columns")
        
        print(f"\n" + "="*80)
        print("EN SHEET DATA (First 5 rows)")
        print("="*80)
        display(df_en.head(5))
        
        print(f"\n" + "="*80)
        print("FR SHEET DATA (First 5 rows)")
        print("="*80)
        display(df_fr.head(5))
        
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        print(f"\nMake sure the sheet names are exactly correct (including spaces, capitalization, etc.)")
```

## STEP 7: Extract Model Information

According to the notes, look for:
1. "1.0 Packages and Kits" marker
2. Rows ABOVE it containing Model Name and Year

```python
if parts_sheet_en and parts_sheet_fr:
    print(f"Searching for model information in: {parts_sheet_en}\n")
    
    df_str = df_en.astype(str)
    
    # Find "1.0 Packages and Kits" marker
    marker_row = None
    for idx, row in df_str.iterrows():
        row_text = " ".join(row.values)
        if "1.0 Packages" in row_text or "Packages and Kits" in row_text:
            marker_row = idx
            print(f"✓ Found '1.0 Packages and Kits' marker at row {idx}")
            break
    
    if marker_row is not None:
        print(f"\nRows ABOVE the marker (rows 0 to {marker_row - 1}):")
        print("="*80)
        print(df_en.iloc[0:marker_row].to_string())
    else:
        print(f"✗ Marker not found. Showing first 20 rows:")
        print("="*80)
        display(df_en.head(20))
```
